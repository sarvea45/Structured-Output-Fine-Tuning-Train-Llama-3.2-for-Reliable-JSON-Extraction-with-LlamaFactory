```python
# 3. Register the Dataset inside LlamaFactory
import json

# Copy the dataset file
!cp "my_repo/data/curated_train.jsonl" "data/structured_invoices.jsonl"

# Update dataset_info.json
with open("data/dataset_info.json", "r") as f:
    dataset_info = json.load(f)

dataset_info["structured_invoices"] = {
    "file_name": "structured_invoices.jsonl",
    "columns": {
        "prompt": "instruction",
        "query": "input",
        "response": "output"
    }
}

with open("data/dataset_info.json", "w") as f:
    json.dump(dataset_info, f, indent=2)

print("Dataset successfully registered as 'structured_invoices'!")
```
